```python
# import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# set default figure size for all plots
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# load the pharmacy sales dataset
df = pd.read_csv('global_pharmacy_sales_2020_2025_daily_dataset (1).csv')

# convert date column to proper datetime type
df['date'] = pd.to_datetime(df['date'])

# check number of rows and columns
print("Shape of dataset:", df.shape)

# show first 5 rows
df.head()
```

```python
# check column data types and non-null counts
df.info()

# check missing values in each column
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
# summary statistics for all numerical columns
df.describe()
```

```python

# UNIVARIATE ANALYSIS - NUMERICAL COLUMNS

# distribution of units sold
plt.hist(df['units_sold'], bins=50, color='skyblue', edgecolor='black')
plt.title('Distribution of Units Sold')
plt.xlabel('Units Sold')
plt.ylabel('Frequency')
plt.show()
```

```python
# distribution of unit price
plt.hist(df['unit_price'], bins=50,edgecolor='black',color='lightgreen')
plt.title('Distribution of Unit Price')
plt.xlabel('Unit Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# distribution of stock level
plt.hist(df['stock_level'], bins=50, color='orange', edgecolor='black')
plt.title('Distribution of Stock Level')
plt.xlabel('Stock Level')
plt.ylabel('Frequency')
plt.show()
```

```python
# distribution of expiry days remaining
plt.hist(df['expiry_days_remaining'], bins=50, edgecolor='black')
plt.title('Distribution of Expiry Days Remaining')
plt.xlabel('Expiry Days Remaining')
plt.ylabel('Frequency')
plt.show()
```

```python
# boxplots to check outliers in main numerical columns
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].boxplot(df['units_sold'])
axes[0].set_title('Units Sold')

axes[1].boxplot(df['unit_price'])
axes[1].set_title('Unit Price')

axes[2].boxplot(df['stock_level'])
axes[2].set_title('Stock Level')

plt.show()
```

```python
# =========================================================
# UNIVARIATE ANALYSIS - CATEGORICAL COLUMNS
# =========================================================

# number of records by region
region_counts = df['region'].value_counts()
plt.bar(region_counts.index, region_counts.values, color='red', edgecolor='black')
plt.title('Number of Records by Region')
plt.xlabel('Region')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# number of records by medicine category
category_counts = df['category'].value_counts()
plt.bar(category_counts.index, category_counts.values, color='yellow')
plt.title('Number of Records by Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
# number of records by individual medicine
medicine_counts = df['medicine'].value_counts()
plt.bar(medicine_counts.index, medicine_counts.values, color='blue')
plt.title('Number of Records by Medicine')
plt.xlabel('Medicine')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# number of records by age group
age_counts = df['age_group'].value_counts()
plt.bar(age_counts.index, age_counts.values)
plt.title('Number of Records by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.show()
```

```python
# proportion of covid vs non-covid period records
covid_counts = df['covid_flag'].value_counts()
plt.pie(covid_counts.values, labels=['Non-Covid', 'Covid'], autopct='%1.1f%%',
        colors=['lightblue', 'salmon'])
plt.title('Covid vs Non-Covid Period Records')
plt.show()
```

```python
# =========================================================
# MULTIVARIATE ANALYSIS
# =========================================================

# correlation matrix between numerical columns
num_cols = ['units_sold', 'unit_price', 'stock_level', 'expiry_days_remaining', 'covid_flag']
corr = df[num_cols].corr()

plt.imshow(corr, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(num_cols)), num_cols, rotation=45)
plt.yticks(range(len(num_cols)), num_cols)
plt.title('Correlation Heatmap of Numerical Columns')

# add correlation values as text on the heatmap
for i in range(len(num_cols)):
    for j in range(len(num_cols)):
        plt.text(j, i, round(corr.iloc[i, j], 2), ha='center', va='center', color='black')

plt.show()
```

```python
# units sold distribution across categories
categories = df['category'].unique()
data_by_cat = [df[df['category'] == cat]['units_sold'] for cat in categories]

plt.boxplot(data_by_cat, labels=categories)
plt.title('Units Sold by Category')
plt.xlabel('Category')
plt.ylabel('Units Sold')
plt.xticks(rotation=45)
plt.show()
```

```python
# average units sold by region
avg_units_region = df.groupby('region')['units_sold'].mean().sort_values(ascending=False)

plt.bar(avg_units_region.index, avg_units_region.values, color='green')
plt.title('Average Units Sold by Region')
plt.xlabel('Region')
plt.ylabel('Average Units Sold')
plt.xticks(rotation=45)
plt.show()
```

```python
# average unit price by category
avg_price_cat = df.groupby('category')['unit_price'].mean().sort_values(ascending=False)

plt.bar(avg_price_cat.index, avg_price_cat.values, color='navy')
plt.title('Average Unit Price by Category')
plt.xlabel('Category')
plt.ylabel('Average Unit Price')
plt.show()
```

```python
# total units sold trend by year
yearly_sales = df.groupby('year')['units_sold'].sum()

plt.plot(yearly_sales.index, yearly_sales.values, marker='o', color='red')
plt.title('Total Units Sold by Year')
plt.xlabel('Year')
plt.ylabel('Total Units Sold')
plt.grid(True)
plt.show()
```

```python
# total units sold trend by month (seasonality check)
monthly_sales = df.groupby('month')['units_sold'].sum()

plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='blue')
plt.title('Total Units Sold by Month')
plt.xlabel('Month')
plt.ylabel('Total Units Sold')
plt.grid(True)
plt.show()
```

```python
# impact of covid period on average units sold
covid_units = df.groupby('covid_flag')['units_sold'].mean()

plt.bar(['Non-Covid', 'Covid'], covid_units.values, color=['gray', 'red'])
plt.title('Average Units Sold: Covid vs Non-Covid')
plt.ylabel('Average Units Sold')
plt.show()
```

```python
# relationship between unit price and units sold
plt.scatter(df['unit_price'], df['units_sold'], alpha=0.3, s=5, color='purple')
plt.title('Unit Price vs Units Sold')
plt.xlabel('Unit Price')
plt.ylabel('Units Sold')
plt.show()
```

```python
# average units sold across region and category combinations
pivot = df.pivot_table(index='region', columns='category', values='units_sold', aggfunc='mean')

plt.imshow(pivot, aspect='auto')
plt.colorbar()
plt.xticks(range(len(pivot.columns)), pivot.columns, rotation=45)
plt.yticks(range(len(pivot.index)), pivot.index)
plt.title('Average Units Sold: Region vs Category')
plt.show()
```

```python
# record counts split by age group and category (stacked bar)
pivot2 = df.pivot_table(index='age_group', columns='category', values='units_sold', aggfunc='count')

pivot2.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.title('Record Count: Age Group vs Category')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.show()
```

```python
# stock level vs expiry days remaining, colored by covid period
colors = df['covid_flag'].map({0: 'blue', 1: 'red'})

plt.scatter(df['stock_level'], df['expiry_days_remaining'], c=colors, alpha=0.3, s=5)
plt.title('Stock Level vs Expiry Days Remaining (Red = Covid Period)')
plt.xlabel('Stock Level')
plt.ylabel('Expiry Days Remaining')
plt.show()
```

```python
# final summary of key findings
print("SUMMARY OF ANALYSIS")
print("Total records:", len(df))
print("Date range:", df['date'].min().date(), "to", df['date'].max().date())
print("Top selling category:", df.groupby('category')['units_sold'].sum().idxmax())
print("Top region by total sales:", df.groupby('region')['units_sold'].sum().idxmax())
print("Most expensive category (avg price):", df.groupby('category')['unit_price'].mean().idxmax())
print("Overall average units sold per record:", round(df['units_sold'].mean(), 2))
```
